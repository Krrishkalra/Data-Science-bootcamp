```python
def counting(filepath):
    with open(filepath,'r') as file:
        lines = file.readlines()
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
        line_count = len(lines)
        return line_count,word_count,char_count
filepath = 'sample.txt'
l_count, w_count, c_count = counting(filepath)
print(f"no. of lines:{l_count}, no. of words:{w_count}, no. of chars:{c_count}")
```

```python
# to write a list of strings to a new file with each string in new line

def lst_string(filepath,lst1):
    with open(filepath,'w') as file:
        for str in lst1:
            file.write(str + '\n')
filepath = 'output.txt'
lst1 = ["Krrish","kalra","faltu"]
lst_string(filepath,lst1)
```

```python
# find a word and replace it with another word in a file

def find_replace(given,replacer,filepath):
    with open(filepath,'r')as file:
        content = file.read()
    new_content = content.replace(given,replacer)
    with open(filepath,'w') as file:
        file.write(new_content)
        
find_replace("data","DATA",'sample.txt')
```

```python
def read_reverse(filename):
    with open(filename, 'r') as file:
        lines = file.readlines()
    for line in reversed(lines):
        print(line.strip())

read_reverse('sample.txt')
```

```python
# merge content of multiple files into a new file

def merge_file(file_lst,output_file):
    with open(output_file,'w') as op_file:
        for file in file_lst:
            with open(file,'r') as ip_file:
                op_file.write(ip_file.read() + '\n')

file = ["sample.txt","output.txt"]
out = 'merge.txt'

merge_file(file,out)
```
