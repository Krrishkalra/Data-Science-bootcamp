```python
# 1. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Set the index to be the first column.
# 2. Create a Pandas DataFrame with columns 'A', 'B', 'C' and index 'X', 'Y', 'Z'. 
# Fill the DataFrame with random integers and access the element at row 'Y' and column 'B'.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1,100, size = (6,4)),columns = ['A','B','C','D'])
print("original data frame: \n",df)

df.set_index("A", inplace = True)
print("data frame after indexing: \n")
print(df)

df2 = pd.DataFrame(np.random.randint(1,20, size = (3,3)),columns = ['A','B','C'], index = ['X','Y','Z'])
print(df2)
```

```python
# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. 
# Add a new column that is the product of the first two columns.
# 2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. 
# Compute the row-wise and column-wise sum.

df1 = pd.DataFrame(np.random.randint(1,20, size = (5,3)))
print("original data frame: \n",df1)

df1['Product'] = df1[0] * df1[1]
print("data frame after adding new column: \n",df1)

df2 = pd.DataFrame(np.random.randint(1,20, size = (4,3)))
print("original data frame: \n",df2)

rows = df2.sum(axis = 1)
cols = df2.sum(axis = 0)
print("row-wise sum: \n",rows)
print("column-wise sum: \n",cols)
```

```python
# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. 
# Introduce some NaN values. Fill the NaN values with the mean of the respective columns.
# 2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. 
# Introduce some NaN values. Drop the rows with any NaN values.

df1 = pd.DataFrame(np.random.randint(1,20, size = (5,3)))
print("original data frame: \n",df1)
df1.iloc[1,1] = np.nan
df1.iloc[3,0] = np.nan
print("data frame after introducing NaN values: \n",df1)
df1.fillna(df1.mean(), inplace = True)
print("data frame after filling NaN values: \n",df1)
print("\n")

df2 = pd.DataFrame(np.random.randint(1,20, size = (6,4)))
print("original data frame: \n",df2)
df2.iloc[2,2] = np.nan
df2.iloc[4,0] = np.nan
print("data frame after introducing NaN values: \n",df2)
df2.dropna(inplace = True)
print("data frame after dropping rows with NaN values: \n",df2)
```
