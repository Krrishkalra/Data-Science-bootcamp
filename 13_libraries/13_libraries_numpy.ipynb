```python
#line plot

import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [1,4,9,16,25]
z = [1,2,3,4,5]

plt.figure(figsize = (9,5))

plt.subplot(1,2,1)
plt.plot(x,y,color = "red")
plt.title("plot 1")

plt.subplot(1,2,2)
plt.plot(y,x,color = "blue")
plt.title("plot 2")

plt.subplot(2,2,3)
plt.plot(x,z,color = "green")
plt.title("plot 3")

plt.subplot(2,2,4)
plt.plot(z,x, color = "orange")
```

```python
# bar plot
w = ['a','b','c','d','e']
v = [5,7,2,4,6]
plt.bar(w,v)
```

```python
# scatter plot
plt.scatter(x,y, color = "blue", marker = "o")
```

```python
# pie chart

labels = ['A', 'B', 'C', 'D', 'E']
sizes = [15, 30, 45, 10, 20]
color = ['red', 'blue', 'green', 'orange', 'purple']
explode = (0,0,0.2,0,0)
plt.pie(sizes, labels = labels, colors = color, autopct = '%1.1f%%', explode = explode)
```

```python
# 1. Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. 
# Replace all the elements in the third column with 1. 
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.
import numpy as np
arr = np.random.randint(1,21,size = (5,5))
print("1: \n Original Array:\n", arr)

arr[:,2] = 1
print("Modified Array:\n", arr)

print("\n 2: ")
arr3 = np.arange(1,17).reshape(4,4)
print("Original Array:\n", arr3)
np.fill_diagonal(arr3,0)
print("Modified Array:\n", arr3)

```

```python
# 1. Create a NumPy array of shape (6, 6) with values from 1 to 36. 
# Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.
# 2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

arr2 = np.arange(1,37).reshape(6,6)
print("1: \n Original Array:\n", arr2)
print("Sub-array:\n", arr2[2:5,1:4])

arr4 = np.arange(1,26).reshape(5,5)
print("2: \n Original Array:\n", arr4)

border_elements = np.concatenate((arr4[0, :], arr4[-1, :], arr4[1:-1, 0], arr4[1:-1, -1]))
print("Border elements:")
print(border_elements)
```

```python
# 1. Create two NumPy arrays of shape (3, 4) filled with random integers. 
# Perform element-wise addition, subtraction, multiplication, and division.
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

arr5 = np.random.randint(1,10,size = (3,4))
arr6 = np.random.randint(2,20,size = (3,4))
arr_sum = arr5 + arr6
arr_dif = arr6 - arr5
arr_prod = arr5*arr6
arr_div = arr6/arr5

print("arr1: ", arr5)
print("\narr2: ", arr6)
print("\nsum : ", arr_sum)
print("\ndifference : ", arr_dif)
print("\nproduct : ", arr_prod)
print("\ndivision : ", arr_div)
```

```python
arr7 = np.arange(1,17).reshape(4,4)
row_sum = arr7.sum(axis = 1)
col_sum = arr7.sum(axis = 0)
print("arr 1= ", arr7)
print("\n")
print("row-wise sum: ", row_sum)
print("column-wise sum: ", col_sum)
```

```python
# 2. Create a NumPy array of shape (3, 3) with values from 1 to 9. 
# Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

arr8 = np.arange(1,10).reshape(3,3)
mean = np.mean(arr8)
std = np.std(arr8)
norm_arr= (arr8-mean)/std

print("Original Array:\n", arr8)
print("Normalized Array:\n", norm_arr)
```

```python
# 1. Create a NumPy array of shape (3, 3) filled with random integers. 
# Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.
# 2. Create a NumPy array of shape (4, 4) filled with random integers. 
# Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

arr9 = np.random.randint(1,17, size = (4,4))
arr10 = np.random.randint(1,17, size = (3,3))
ar1 = np.random.randint(1,20, size = (4,))
ar2 = np.random.randint(1,20, size = (3,))
print("original: \n", arr9)
print("1D: ",ar1)

result = arr9 - ar1[:,np.newaxis]
print("after subtraction: \n",result)

result_sum = arr10 + ar2
print("original: \n", arr10)
print("1D: ",ar2)
print("after addition: \n",result_sum)
```
