```python
# Create a set with the first 10 positive integers. Print the set.
# Add the number 11 to the set. Then remove the number 1 from the set. 
# Print the modified set.

st = set(range(1,11))
print(st)
st.add(11)
print(st)
st.remove(1)
print(st)
```

```python
# Create two sets: one with the first 5 positive integers and another with the first 5 even integers. 
# Perform and print the results of 
# union, intersection, difference, and symmetric difference operations on these sets.

st1 = set(range(1,6))
st2 = set(range(2,11,2))
print(st1)
print(st2)
st3 = st1.union(st2)
st4 = st1.intersection(st2)
print(st3)
print(st4)
print(st1.difference(st2))
print(st1.symmetric_difference(st2))
```

```python
# Create a new set containing the squares of the first 10 positive integers 
# using a set comprehension. Print the new set.

square = {x**2 for x in range(1,11)}
squares = sorted(square)
print(squares)
```

```python
# Create a new set containing only the even numbers from the set created 
# in Assignment 1 using a set comprehension. Print the new set.

st = set(range(1,11))
stnew = {x for x in st if x%2 == 0}
print(stnew)
```
