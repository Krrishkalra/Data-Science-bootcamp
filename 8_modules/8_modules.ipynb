```python
#Use the `datetime` module to print the current date, 
# calculate the date 100 days from today, and determine the day of the week for a given date.

import datetime as dt
today = dt.date.today()
print(today)

future = today + dt.timedelta(days=100)
print(future)

given = dt.date(2025,12,6)
print(f"week of {given} date is {given.strftime('%A')}")
```

```python
# Use the `random` module to generate a list of 5 random numbers 
# between 1 and 50 and shuffle the elements of a list.

import random 
lst = []
for i in range(0,5):
    lst.append(random.randint(1,50))
print(lst)

random.shuffle(lst)
print(lst)
```
