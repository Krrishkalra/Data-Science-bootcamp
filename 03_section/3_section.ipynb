```python
# Write a program that asks the user to input numbers until they input 0. 
# The program should print the sum of all the input numbers.
sum = 0
while True:
    num = int(input("enter a number (0 to stop): "))
    if num == 0:
        break
    sum = sum + num
print(f"sum is {sum}")
```

```python
# Write a program that asks the user to input a number 
# and prints all the even numbers from 1 to that number using a for loop.

num = int(input("enter a number : "))
for i in range(1,num+1):
    if i%2 == 0:
        print(i)
```

```python
# Write a program that calculates the sum of the digits of a number
#  input by the user using a while loop.

num = int(input("enter number: "))
sum = 0
while(num > 0):
    x = num%10
    sum = sum + x
    num = num//10

print(sum)
```

```python
# Write a program that checks if a number 
# input by the user is a prime number using a for loop.

isPrime = True
num = int(input("enter a number "))
for i in range(2,int((num**0.5)+1)):
    if num%i == 0:
        isPrime = False
        break
    else:
        isPrime = True 
        
if isPrime == True:
    print(num, "is prime")
else:
    print(num, "is not prime")
```

```python
# Write a program that prints the first n Fibonacci numbers,
# where n is input by the user.

n = int(input("enter num for fibo: "))
count = 0
a,b = 0,1
while count < n:
    print(a)
    a,b = b, a+b
    count += 1
    
```
