```python
import sqlite3 
```

```python
# creating connection
connection = sqlite3.connect("example.db")
connection
```

```python
#creating cursor
cursor = connection.cursor()

# creating table
cursor.execute('''
CREATE TABLE IF NOT EXISTS employees (
    id Integer Primary Key,
    name Text NOT NULL,
    age INTEGER,
    dept Text
    )
''')

connection.commit()

```

```python
cursor.execute('''
    INSERT INTO employees(id, name, age, dept)
        Values(2, 'John Doe', 30, 'HR')
        
''')

connection.commit()
```

```python
cursor.execute('''
    Select * from employees
''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# update data into table
cursor.execute('''
    UPDATE employees
    Set name = 'Jane Doe',
    age = 28,
    dept = 'Finance'
    where id = 2
''')
connection.commit()
```

```python
employees = [
    (3, 'Alice Smith', 25, 'Marketing'),
    (4, 'Bob Johnson', 35, 'Sales'),
    (5, 'Charlie Brown', 28, 'IT'),
    (6, 'David Wilson', 32, 'HR')
]

cursor.executemany('''
    Insert into employees(id, name, age, dept)
    values(?, ?, ?, ?)
''', employees)

connection.commit()
cursor.execute('''
    Select * from employees
''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Write a Python function to fetch and display all employees from a specific department.

def fetch_employees_by_dept(dept):
    cursor.execute('''Select * from employees where dept = ?''', (dept,))
    rows = cursor.fetchall()
    for row in rows:
        print(row)
        
fetch_employees_by_dept('Finance')
```

```python
# Write a Python function to delete an employee from the `employees` table based on their `id`.
def delete_emp(emp_id):
    cursor.execute('''Delete from employees where id = ?''',(emp_id,))
    connection.commit()
    cursor.execute('''
    Select * from employees''')
    rows = cursor.fetchall()
    for row in rows:
        print(row)
        
delete_emp(3)
```

```python
# 2. Write a Python function to update the age of multiple employees in a single transaction. 
# Ensure that if any update fails, none of the updates are committed.

def update_ages(new_age,old_age):
    try:
        connection = sqlite3.connect("example.db")
        cursor = connection.cursor()
        cursor.execute('''Update employees set age = ? where age = ?''',(new_age,old_age))
        connection.commit()
        print("Ages updated successfully")
        
    except Exception as e:
        connection.rollback()
        print("An error occurred:", e)
    finally:
        cursor.execute('''Select * from employees''')
        rows = cursor.fetchall()
        for row in rows:
            print(row)
        
update_ages(28,30)
```
