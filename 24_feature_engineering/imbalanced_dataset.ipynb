### FEATURE ENGINEERING
### IMBALANCED DATASET 
### PERFORMING Up Scaling AND DOWN Scaling

```python
import numpy as np
import pandas as pd

np.random.seed(123)

n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0]*n_class_0
})
class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1]*n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
df.tail()
```

```python
df['target'].value_counts()
```

### UP SCALING

```python
df_minor = df[df['target']==1]
df_major = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minor_upsampled = resample(df_minor, replace=True, # sample with replacement
         n_samples = len(df_major),
         random_state = 42
         )
```

```python
df_minor_upsampled.head()
```

```python
df_upsampled = pd.concat([df_major,df_minor_upsampled])
df_upsampled['target'].value_counts()
```

### DOWN SCALING

```python
df_minor = df[df['target']==1]
df_major = df[df['target']==0]
from sklearn.utils import resample
df_major_downsampled = resample(df_major, replace=False, # sample with replacement
         n_samples = len(df_minor),
         random_state = 42
         )
df_major_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minor, df_major_downsampled])
df_downsampled
```

```python
df_downsampled['target'].value_counts()
```
