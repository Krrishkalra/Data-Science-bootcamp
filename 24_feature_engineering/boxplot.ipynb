### creating box plot using 5 number summary ( minimum, maximum , Q1, Q3)
we will also learn about 
IQR

```python
import numpy as np
```

```python
lst_marks = [45,32,56,75,89,54,32,89,90,87,67,54,45,98,67,74,150,154,160]
min,Q1,median,Q3,max = np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
min,Q1,median,Q3,max
```

```python
IQR = Q3-Q1
print(IQR)
```

```python
lower_fence = Q1 - 1.5*(IQR)
higher_fence = Q3 + 1.5*(IQR)
```

```python
lower_fence
```

```python
higher_fence
```

```python
import seaborn as sns
sns.boxplot(lst_marks)
```
