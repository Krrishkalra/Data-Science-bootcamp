```python
try:
    num = int(input("enter no.: "))
    result = 10/num
    
except ValueError:
    print("enter a valid number")
except ZeroDivisionError:
    print("division by zero is not allowed")
except Exception as ex:
    print(ex)
else: 
    print(result)
finally:
    print("execution completed")
```

```python
# wap a function to read content of file and handle file not found exception 

def reading(file_path):
    try:
        with open(file_path,'r') as file:
            content = file.readlines()
            return content
    except FileNotFoundError as e:
        print(f"Error : {e}")
    finally:
        try:    
            file.close()
        except NameError:
            pass

reading('output2.txt')
```

```python
# exception handling in dictionary access

def dict1(d,key):
    try:
        value = d[key]
    except KeyError:
        print(f"{key} not found in the dictionary")
        value  = None
    finally:
        print("execution completed") 
    return value

dict1({1:2,2:3,3:4,5:6},3)
        
```

```python
# nested try exception handling with valueError and ZerodivisionError

def nestedError(s):
    try:
        try:
            num = int(s)
        except ValueError as e:
            print(f"Error: {e}")
            num = None
        finally:
            print("conversion attempt complete")
        if num is not None:
            try:
                result = 10/num
            except ZeroDivisionError as e1:
                print(f"Error: {e1}")
                result = None
            finally:
                print("division attempt complete")
                return result
    finally:
        print("overall execution completed")       
nestedError('5')
                
```
